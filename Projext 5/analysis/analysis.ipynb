### 📝 Task 1: What are Ensemble Learning Methods and why are they effective?

Ensemble Learning is a machine learning technique that combines multiple models to make better predictions. It improves accuracy, reduces errors, and provides more reliable results than using a single model.

**📌 Conclusion:**

Ensemble Learning improves prediction performance by combining multiple models and reducing overall errors.

### 📝 Task 2: Explain the difference between Bagging, Boosting, and Voting.

- **Bagging:** Trains multiple models independently on different samples of data and combines their predictions.
- **Boosting:** Trains models sequentially, where each new model focuses on correcting previous errors.
- **Voting:** Combines predictions from multiple machine learning models using majority voting or average probabilities.

**📌 Conclusion:**

Bagging reduces variance, Boosting improves weak models, and Voting combines multiple predictions to achieve better results.

### 📝 Task 3: What is Bias–Variance Trade-off, and how do ensemble methods address it?

Bias–Variance Trade-off is the balance between underfitting and overfitting in machine learning models. Bagging helps reduce variance, while Boosting helps reduce bias. Ensemble methods improve overall prediction performance by balancing both.

**📌 Conclusion:**

Ensemble methods help balance bias and variance to improve model accuracy and reduce prediction errors.

### 📝 Task 4: Explain Voting Classifier and Stacking Ensemble.

- **Voting Classifier:** Combines predictions from multiple machine learning models. It uses majority voting or average probabilities to produce the final prediction.
- **Stacking Ensemble:** Combines multiple base models using another model called a Meta-Model. The meta-model learns from the predictions of base models to make the final prediction.

**📌 Conclusion:**

Voting and Stacking combine multiple models to improve prediction performance and produce more reliable results.

### 📝 Task 5: Compare AdaBoost and XGBoost conceptually.

**AdaBoost:** AdaBoost is a boosting technique that trains models sequentially. It gives more importance to incorrectly predicted data points so that the next model can focus on previous mistakes.

**XGBoost:** XGBoost is an advanced boosting algorithm that uses gradient boosting with regularization. It improves prediction accuracy, controls overfitting, and provides optimized training.

| AdaBoost | XGBoost |
|---|---|
| Focuses on misclassified samples. | Focuses on reducing prediction errors using gradients. |
| Uses sample weights. | Uses gradient and second-order information. |
| Simpler boosting algorithm. | More advanced and optimized algorithm. |
| Less control over regularization. | Supports regularization. |

**📌 Conclusion:**

AdaBoost focuses on correcting misclassified samples, while XGBoost uses gradient boosting and regularization for optimized model performance.

## 🎯 Part A: Final Conclusion

Ensemble Learning is an effective machine learning approach that combines multiple models to improve prediction accuracy and reliability. Techniques such as Bagging, Boosting, Voting, and Stacking help reduce errors and improve model performance. AdaBoost and XGBoost are important boosting algorithms used to build powerful predictive models.

## 🔹 Task 6: Identify Features and Target Variables

```python
# 🔍 Task 6: Identify Features and Target Variables

# Define Classification Target
classification_target = 'completion_status'

# Define Regression Target
regression_target = 'final_score'

# Remove ID, Date and Target Columns
features = df.drop(
    columns=[
        'student_id',
        'course_start_date',
        'completion_status',
        'final_score'
    ]
)

# Define Target Variables
X = features
y_classification = df[classification_target]
y_regression = df[regression_target]

# Display Features and Targets
print("Total Features:", X.shape[1])

print("\nFeature Columns:")
print(X.columns.tolist())

print("\nClassification Target:", classification_target)
print("Regression Target:", regression_target)

print("\nClassification Target Distribution:")
print(y_classification.value_counts())

print("\nRegression Target Statistics:")
print(y_regression.describe())
```

### 📌 Task 6: Conclusion

The dataset contains demographic, engagement, and assessment-related features. The `completion_status` column is selected as the classification target, while `final_score` is selected as the regression target. The student ID and course start date are excluded from the input features.

## 🔹 Task 7: Perform Train–Test Split

```python
# ✂️ Task 7: Train-Test Split

# Classification Train-Test Split
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)

# Regression Train-Test Split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_regression,
    test_size=0.20,
    random_state=42
)

# Display Classification Split
print("Classification Dataset")
print("Training Data:", X_train_cls.shape)
print("Testing Data:", X_test_cls.shape)

# Display Regression Split
print("\nRegression Dataset")
print("Training Data:", X_train_reg.shape)
print("Testing Data:", X_test_reg.shape)
```

### 📌 Task 7: Conclusion

The dataset is divided into 80% training data and 20% testing data for both classification and regression tasks. The classification split uses stratification to maintain the target class distribution. A random state of 42 ensures reproducible results.

## 🔹 Task 8: Apply Basic Preprocessing

```python
# ⚙️ Task 8: Basic Data Preprocessing

# Identify Numerical and Categorical Columns
numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()

# Numerical Preprocessing
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical Preprocessing
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Create Preprocessor
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Classification Preprocessing
X_train_cls_processed = preprocessor.fit_transform(X_train_cls)
X_test_cls_processed = preprocessor.transform(X_test_cls)

# Regression Preprocessing
X_train_reg_processed = preprocessor.fit_transform(X_train_reg)
X_test_reg_processed = preprocessor.transform(X_test_reg)

# Display Results
print("Numerical Features:", len(numeric_features))
print("Categorical Features:", len(categorical_features))

print("\nClassification Processed Data:")
print("Training Shape:", X_train_cls_processed.shape)
print("Testing Shape:", X_test_cls_processed.shape)

print("\nRegression Processed Data:")
print("Training Shape:", X_train_reg_processed.shape)
print("Testing Shape:", X_test_reg_processed.shape)

print("\nPreprocessing Completed Successfully!")
```

### 📌 Task 8: Conclusion

Basic preprocessing is successfully applied to the dataset. Missing numerical values are handled using median imputation, while categorical missing values are handled using mode imputation. StandardScaler is used for numerical scaling, and OneHotEncoder converts categorical features into numerical format.

## 📝 Task 9: Implement Bagging Classifier

```python
# 🌳 Task 9: Implement Bagging Classifier

from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier

# Create Bagging Classifier
bagging_classifier = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=42),
    n_estimators=100,
    max_samples=0.8,
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

# Train Model
bagging_classifier.fit(
    X_train_cls_processed,
    y_train_cls
)

# Make Predictions
y_pred_bagging_cls = bagging_classifier.predict(
    X_test_cls_processed
)

# Evaluate Model
bagging_accuracy = accuracy_score(
    y_test_cls,
    y_pred_bagging_cls
)

bagging_precision = precision_score(
    y_test_cls,
    y_pred_bagging_cls,
    zero_division=0
)

bagging_recall = recall_score(
    y_test_cls,
    y_pred_bagging_cls,
    zero_division=0
)

bagging_f1 = f1_score(
    y_test_cls,
    y_pred_bagging_cls,
    zero_division=0
)

# Display Results
print("Bagging Classifier Results")
print("--------------------------")
print("Accuracy:", round(bagging_accuracy, 4))
print("Precision:", round(bagging_precision, 4))
print("Recall:", round(bagging_recall, 4))
print("F1 Score:", round(bagging_f1, 4))
```

### 📌 Task 9: Conclusion

The Bagging Classifier was successfully implemented using 100 Decision Trees. The model combines predictions from multiple trees to predict whether students will complete their courses. Its performance was evaluated using Accuracy, Precision, Recall, and F1 Score.

## 📝 Task 10: Implement Bagging Regressor

```python
# 📊 Task 10: Implement Bagging Regressor

# Create Bagging Regressor
bagging_regressor = BaggingRegressor(
    estimator=DecisionTreeRegressor(random_state=42),
    n_estimators=100,
    max_samples=0.8,
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

# Train Model
bagging_regressor.fit(
    X_train_reg_processed,
    y_train_reg
)

# Make Predictions
y_pred_bagging_reg = bagging_regressor.predict(
    X_test_reg_processed
)

# Evaluate Model
bagging_mae = mean_absolute_error(
    y_test_reg,
    y_pred_bagging_reg
)

bagging_mse = mean_squared_error(
    y_test_reg,
    y_pred_bagging_reg
)

bagging_rmse = np.sqrt(bagging_mse)

bagging_r2 = r2_score(
    y_test_reg,
    y_pred_bagging_reg
)

# Display Results
print("Bagging Regressor Results")
print("-------------------------")
print("MAE:", round(bagging_mae, 4))
print("MSE:", round(bagging_mse, 4))
print("RMSE:", round(bagging_rmse, 4))
print("R2 Score:", round(bagging_r2, 4))
```

### 📌 Task 10: Conclusion

The Bagging Regressor was successfully implemented using 100 Decision Tree Regressors. It predicts students' final scores by combining predictions from multiple trees. Its performance was evaluated using MAE, MSE, RMSE, and R² Score.
