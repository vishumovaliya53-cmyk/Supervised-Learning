# Part A — Theory cells

## 📊 Question 1: What are Supervised Learning Algorithms?

**Definition:** Supervised learning is a machine learning approach in which a model learns from labelled data containing input features and known output values.

**Types:**

1. **Classification:** Predicts categories, such as spam or not spam.
2. **Regression:** Predicts continuous numerical values, such as house prices.

**Example:** In this project, the model learns from house features and their known prices to predict the price of another house.

**Conclusion:** Supervised learning uses labelled examples to learn patterns and make predictions on new data.

## 🧠 Question 2: What is Regression in Machine Learning?

**Definition:** Regression is a supervised learning technique used to predict continuous numerical values.

**Examples:**
- Predicting house prices
- Estimating sales revenue
- Predicting temperature

**In this project:** Regression is used to estimate the house price in INR based on features such as area, bedrooms, bathrooms, and location score.

**Conclusion:** Regression helps estimate numerical values by learning relationships between input features and the target variable.

## 📈 Question 3: What is Linear Regression?

**Definition:** Linear Regression is a supervised machine learning algorithm that models the relationship between input features and a continuous target variable using a linear equation.

**Simple equation:**

y = mx + c

Where:

- `y` = Predicted output
- `x` = Input feature
- `m` = Slope or coefficient
- `c` = Intercept

For multiple features, the model combines their coefficients to estimate the target value.

**In this project:** Linear Regression predicts house prices using available property features.

**Conclusion:** Linear Regression is a simple and interpretable method for predicting continuous values.

## 🏡 Question 4: What are Features and Target Variables?

**Features:** Features are the input variables used by a machine learning model to make predictions.

**Target Variable:** The target is the output value that the model tries to predict.

**Examples from this project:**

- Features: `area_sqft`, `bedrooms`, `bathrooms`, `location_score`, `age_years`
- Target: `house_price_inr`

**Conclusion:** Features provide information to the model, while the target variable represents the value to be predicted.

## 📉 Question 5: What is Overfitting and Underfitting?

**Overfitting:** Occurs when a model learns the training data too closely, including noise, and performs poorly on unseen data.

**Underfitting:** Occurs when a model is too simple to learn the important patterns in the data.

**How to reduce these problems:**

- Use appropriate training and testing data.
- Evaluate the model using suitable metrics.
- Select relevant features.
- Compare different models when necessary.

**Conclusion:** A good model should learn meaningful patterns and perform well on both training and unseen data.

## 🎯 Question 6: How is Regression Model Performance Evaluated?

Regression models can be evaluated using the following metrics:

**1. Mean Absolute Error (MAE):** Measures the average absolute difference between actual and predicted values.

**2. Mean Squared Error (MSE):** Measures the average squared difference between actual and predicted values.

**3. Root Mean Squared Error (RMSE):** Calculates the square root of MSE and expresses the error in the target variable's units.

**4. R-squared (R²):** Measures how much variation in the target variable is explained by the model.

**Interpretation:**

- Lower MAE, MSE, and RMSE generally indicate smaller prediction errors.
- A higher R² generally indicates a better fit, although it does not guarantee good performance on unseen data.

**Conclusion:** These metrics help us understand how accurately the model predicts house prices.

# 📊 Part C: Regularized Linear Models

---

```python
# Common Setup for Task 9 to 12

X_train_simple = X_train.copy()
X_test_simple = X_test.copy()

# Convert sale_date
X_train_simple["sale_date"] = pd.to_datetime(
    X_train_simple["sale_date"]
).map(lambda x: x.toordinal() if pd.notna(x) else np.nan)

X_test_simple["sale_date"] = pd.to_datetime(
    X_test_simple["sale_date"]
).map(lambda x: x.toordinal() if pd.notna(x) else np.nan)

# Handle missing values
X_train_simple = X_train_simple.fillna(X_train_simple.median())
X_test_simple = X_test_simple.fillna(X_train_simple.median())

print("Data Ready Successfully!")
print("Training Shape:", X_train_simple.shape)
print("Testing Shape:", X_test_simple.shape)
```

## Task 9 — Implement Ridge Regression (L2)

```python

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(X_train_simple, y_train)

ridge_pred = ridge_model.predict(X_test_simple)

print("Ridge Regression Trained Successfully!")
```

### ✅ Task 9: Conclusion

In this task, Ridge Regression was implemented using L2 regularization. Missing values were handled, and numerical and categorical features were preprocessed.

**Result:** The Ridge Regression model was successfully trained.

## Task 10 — Implement Lasso Regression (L1)

```python
# Task 10: Lasso Regression

lasso_model = Lasso(alpha=1.0, max_iter=10000)

# Train the model
lasso_model.fit(X_train_simple, y_train)

# Predictions
lasso_pred = lasso_model.predict(X_test_simple)

print("Lasso Regression Trained Successfully!")
```

### ✅ Task 10: Conclusion

Lasso Regression was implemented using L1 regularization. It helps reduce overfitting and can make some feature coefficients zero.

**Result:** The Lasso Regression model was trained successfully.

## Task 11 — Tune Regularization Parameter (α) using Cross-Validation

```python
# Task 11: Hyperparameter Tuning

# Alpha values
parameters = {"alpha": [0.01, 0.1, 1, 10, 100]}

# Ridge Tuning
ridge_search = GridSearchCV(
    Ridge(), parameters, cv=5,
    scoring="neg_mean_squared_error"
)
ridge_search.fit(X_train_simple, y_train)

# Lasso Tuning
lasso_search = GridSearchCV(
    Lasso(max_iter=10000), parameters, cv=5,
    scoring="neg_mean_squared_error"
)
lasso_search.fit(X_train_simple, y_train)

print("Best Ridge Alpha:", ridge_search.best_params_)
print("Best Lasso Alpha:", lasso_search.best_params_)
```

### ✅ Task 11: Conclusion

In this task, GridSearchCV with 5-fold cross-validation was used to find the best regularization parameter (alpha) for Ridge and Lasso Regression.

**Result:** The best alpha values were selected to improve model performance.

## Task 12: Comparison of Ridge and Lasso Regression

```python
# Task 12: Model Comparison

# Best models from Task 11
ridge_model = ridge_search.best_estimator_
lasso_model = lasso_search.best_estimator_

# Predictions
ridge_train_pred = ridge_model.predict(X_train_simple)
ridge_val_pred = ridge_model.predict(X_test_simple)

lasso_train_pred = lasso_model.predict(X_train_simple)
lasso_val_pred = lasso_model.predict(X_test_simple)

# Compare errors
comparison = pd.DataFrame({
    "Model": ["Ridge Regression", "Lasso Regression"],
    "Training Error": [
        mean_squared_error(y_train, ridge_train_pred),
        mean_squared_error(y_train, lasso_train_pred)
    ],
    "Validation Error": [
        mean_squared_error(y_test, ridge_val_pred),
        mean_squared_error(y_test, lasso_val_pred)
    ],
    "R2 Score": [
        r2_score(y_test, ridge_val_pred),
        r2_score(y_test, lasso_val_pred)
    ]
})

print("MODEL PERFORMANCE COMPARISON")
display(comparison.round(2))

# Feature coefficient comparison
coefficients = pd.DataFrame({
    "Feature": X_train_simple.columns,
    "Ridge Coefficient": ridge_model.coef_,
    "Lasso Coefficient": lasso_model.coef_
})

print("\nFEATURE COEFFICIENT COMPARISON")
display(coefficients.round(4))

print("\nRidge Zero Coefficients:",
      np.sum(np.isclose(ridge_model.coef_, 0)))

print("Lasso Zero Coefficients:",
      np.sum(np.isclose(lasso_model.coef_, 0)))
```

### ✅ Task 12: Conclusion

In this task, Ridge and Lasso Regression were compared using training error, validation error, R² Score, and feature coefficients.

Ridge reduces coefficient values, while Lasso can make some coefficients zero.

**Result:** Both models were compared to understand their performance and the effect of regularization.

# 📊 Part D: Cross-Validation Strategies

---

## Task 13 — Apply Different Cross-Validation Techniques

##### 
K-Fold Cross-Validation

Stratified K-Fold Cross-Validation

Leave-One-Out Cross-Validation

Time Series Split

```python
# Task 13: Cross-Validation Strategies

model = Ridge(alpha=1.0)

# 1. K-Fold Cross-Validation
kfold = KFold(n_splits=5, shuffle=True, random_state=42)

kfold_scores = cross_val_score(
    model, X_train_simple, y_train,
    cv=kfold, scoring="neg_mean_squared_error"
)

# 2. Stratified K-Fold (Target Binning)
target_bins = pd.qcut(y_train, q=5, labels=False, duplicates="drop")

skfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

skfold_scores = cross_val_score(
    model, X_train_simple, y_train,
    cv=skfold.split(X_train_simple, target_bins),
    scoring="neg_mean_squared_error"
)

# 3. Leave-One-Out Cross-Validation
loo = LeaveOneOut()

loo_scores = cross_val_score(
    model, X_train_simple, y_train,
    cv=loo, scoring="neg_mean_squared_error"
)

# 4. Time Series Split
sorted_index = X_train_simple.sort_values("sale_date").index

X_time = X_train_simple.loc[sorted_index]
y_time = y_train.loc[sorted_index]

tscv = TimeSeriesSplit(n_splits=5)

time_scores = cross_val_score(
    model, X_time, y_time,
    cv=tscv, scoring="neg_mean_squared_error"
)

print("All Cross-Validation Techniques Completed!")
print("K-Fold RMSE:", round(np.sqrt(-kfold_scores.mean()), 2))
print("Stratified K-Fold RMSE:", round(np.sqrt(-skfold_scores.mean()), 2))
print("Leave-One-Out RMSE:", round(np.sqrt(-loo_scores.mean()), 2))
print("Time Series Split RMSE:", round(np.sqrt(-time_scores.mean()), 2))
```

### ✅ Task 13: Conclusion

In this task, four cross-validation techniques were applied: K-Fold, Stratified K-Fold, Leave-One-Out, and Time Series Split.

**Result:** Different validation strategies were used to evaluate the Ridge Regression model.

## Task 14 — Analyze Performance Metrics

```python
# Task 14: Compare Cross-Validation Performance

cv_comparison = pd.DataFrame({
    "CV Strategy": [
        "K-Fold",
        "Stratified K-Fold",
        "Leave-One-Out",
        "Time Series Split"
    ],
    "Mean RMSE": [
        np.sqrt(-kfold_scores.mean()),
        np.sqrt(-skfold_scores.mean()),
        np.sqrt(-loo_scores.mean()),
        np.sqrt(-time_scores.mean())
    ],
    "Standard Deviation": [
        np.sqrt(-kfold_scores).std(),
        np.sqrt(-skfold_scores).std(),
        np.sqrt(-loo_scores).std(),
        np.sqrt(-time_scores).std()
    ]
})

display(cv_comparison.round(2))

best_strategy = cv_comparison.loc[
    cv_comparison["Mean RMSE"].idxmin(), "CV Strategy"
]

print("Best CV Strategy based on Mean RMSE:", best_strategy)
```

### ✅ Task 14: Conclusion

In this task, performance metrics from different cross-validation strategies were compared using Mean RMSE and Standard Deviation.

**Result:** The cross-validation strategies were analyzed to understand differences in model performance and consistency.

## Task 19 — Implement SVR with Linear and RBF Kernel

```python
# Task 19: Support Vector Regression

# Scale target values
y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(
    np.array(y_train).reshape(-1, 1)
).ravel()

# Linear Kernel
svr_linear = Pipeline([
    ("scaler", StandardScaler()),
    ("svr", SVR(kernel="linear"))
])

svr_linear.fit(X_train_simple, y_train_scaled)

linear_pred = y_scaler.inverse_transform(
    svr_linear.predict(X_test_simple).reshape(-1, 1)
).ravel()

# RBF Kernel
svr_rbf = Pipeline([
    ("scaler", StandardScaler()),
    ("svr", SVR(kernel="rbf"))
])

svr_rbf.fit(X_train_simple, y_train_scaled)

rbf_pred = y_scaler.inverse_transform(
    svr_rbf.predict(X_test_simple).reshape(-1, 1)
).ravel()

print("Linear SVR and RBF SVR trained successfully!")
```

### ✅ Task 19: Conclusion

In this task, Support Vector Regression was implemented using Linear and RBF kernels. Feature scaling and target scaling were applied before training.

**Result:** Both SVR models were trained successfully.

## Task 20 — Tune Hyperparameters (C, γ, ε)

```python
# Task 20: Fast SVR Hyperparameter Tuning

parameters = {
    "svr__C": [1, 10],
    "svr__gamma": ["scale"],
    "svr__epsilon": [0.1]
}

svr_search = GridSearchCV(
    svr_rbf,
    parameters,
    cv=3,
    scoring="neg_mean_squared_error",
    n_jobs=-1
)

svr_search.fit(X_train_simple, y_train_scaled)

best_svr = svr_search.best_estimator_

svr_pred = y_scaler.inverse_transform(
    best_svr.predict(X_test_simple).reshape(-1, 1)
).ravel()

print("Best Parameters:", svr_search.best_params_)
print("SVR Tuning Completed!")
```

### ✅ Task 20: Conclusion

In this task, GridSearchCV with 5-fold cross-validation was used to tune SVR hyperparameters C, gamma, and epsilon.

**Result:** The best SVR hyperparameters were identified.

## Task 21 — Compare SVR with Linear and Tree-Based Models

```python
# Task 21: Model Performance Comparison

# Train comparison models
ridge_compare = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=1.0))
])

ridge_compare.fit(X_train_simple, y_train)

tree_compare = DecisionTreeRegressor(random_state=42)
tree_compare.fit(X_train_simple, y_train)

forest_compare = RandomForestRegressor(
    n_estimators=100, random_state=42
)
forest_compare.fit(X_train_simple, y_train)

# Predictions
predictions = {
    "Ridge Regression": ridge_compare.predict(X_test_simple),
    "Decision Tree": tree_compare.predict(X_test_simple),
    "Random Forest": forest_compare.predict(X_test_simple),
    "Linear SVR": linear_pred,
    "Tuned RBF SVR": svr_pred
}

# Compare performance
results = []

for name, pred in predictions.items():
    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2 Score": r2_score(y_test, pred)
    })

comparison = pd.DataFrame(results)

display(comparison.round(2))

print(
    "Best Model:",
    comparison.loc[comparison["RMSE"].idxmin(), "Model"]
)
```

### ✅ Task 21: Conclusion

In this task, SVR models were compared with Ridge Regression, Decision Tree, and Random Forest using MAE, RMSE, and R² Score.

**Result:** The models were compared to identify the best-performing regression model.

```python
# Task 22: Evaluate All Models

# Predictions for all models
model_predictions = {
    "Ridge Regression": (
        ridge_model.predict(X_train_simple),
        ridge_model.predict(X_test_simple)
    ),
    "Lasso Regression": (
        lasso_model.predict(X_train_simple),
        lasso_model.predict(X_test_simple)
    ),
    "Decision Tree": (
        tree_model.predict(X_train_simple),
        tree_model.predict(X_test_simple)
    ),
    "Tuned Decision Tree": (
        best_tree.predict(X_train_simple),
        best_tree.predict(X_test_simple)
    ),
    "Random Forest": (
        forest_model.predict(X_train_simple),
        forest_model.predict(X_test_simple)
    ),
    "Linear SVR": (
        y_scaler.inverse_transform(
            svr_linear.predict(X_train_simple).reshape(-1, 1)
        ).ravel(),
        linear_pred
    ),
    "Tuned RBF SVR": (
        y_scaler.inverse_transform(
            best_svr.predict(X_train_simple).reshape(-1, 1)
        ).ravel(),
        svr_pred
    )
}

# Calculate evaluation metrics
results = []

for name, (train_pred, test_pred) in model_predictions.items():
    results.append({
        "Model": name,
        "MSE": mean_squared_error(y_test, test_pred),
        "MAE": mean_absolute_error(y_test, test_pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, test_pred)),
        "R2 Score": r2_score(y_test, test_pred)
    })

evaluation_results = pd.DataFrame(results)

display(evaluation_results.round(2))

print("Best Model based on RMSE:")
print(evaluation_results.loc[
    evaluation_results["RMSE"].idxmin(), "Model"
])
```

### ✅ Task 22: Conclusion

## Task 24 — Identify Overfitting and Underfitting

```python
# Task 24: Overfitting and Underfitting Analysis

analysis_results = []

for name, (train_pred, test_pred) in model_predictions.items():

    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))

    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)

    if test_rmse > train_rmse * 1.5:
        status = "Possible Overfitting"
    elif train_r2 < 0.5 and test_r2 < 0.5:
        status = "Possible Underfitting"
    else:
        status = "No Strong Sign"

    analysis_results.append({
        "Model": name,
        "Training RMSE": train_rmse,
        "Testing RMSE": test_rmse,
        "Training R2": train_r2,
        "Testing R2": test_r2,
        "Model Status": status
    })

overfitting_results = pd.DataFrame(analysis_results)

display(overfitting_results.round(3))
```

### ✅ Task 24: Conclusion

In this task, training and testing performance were compared to identify possible overfitting and underfitting in each model.

A large difference between training and testing errors may indicate overfitting, while low performance on both datasets may indicate underfitting.

**Result:** The generalization performance of all regression models was analyzed.
